NumPy

```python
import numpy as np

# Array
a = np.array([1,2,3])

a.shape
a.ndim
a.dtype

# Array Creation
np.zeros((3,4))

np.ones((3,4))

np.eye(3)

np.random.rand(3,4)

np.arange(10)

np.linspace(0,1,100)

# Reshape
a.reshape(2,3)

a.flatten()

# Statistics
np.mean(a)

np.std(a)

np.median(a)

np.min(a)

np.max(a)

np.sum(a)

# Axis
a.mean(axis=0) # mean columns
a.mean(axis=1) # mean rows

# Boolean Masking
a[a > 5]

a[(a > 5) & (a < 10)]

# Matrix Operations

A @ B # Matrix Multiplication equivalent to np.dot
#A.shape = (m,n)
# B.shape = (n,k)
# Result shape = (m,k)

np.dot(A,B) 

A.T

np.linalg.inv(A) # Inverse of a matrix

np.linalg.eig(A) # Eigenvalues and Eigenvectors

# Arg Functions
np.argmax(a)

np.argmin(a)

np.argsort(a)

np.argwhere(a == 5)
```

Pandas

```python
import pandas as pd

# Load
df = pd.read_csv("data.csv")

df = pd.read_excel("data.xlsx")

# Inspect
df.head()

df.tail()

df.info()

df.describe()

df.shape

# Columns

df.columns

df["age"]

df[["age","salary"]]

# Indexing
df.iloc[0] # First row

df.iloc[0:5] # First 5 rows

df.loc[df["name"] == "Hodaya"] # Row where name is Hodaya


# Filtering
df[df["age"] > 30]

df[(df["age"] > 30) &
   (df["salary"] > 10000)]

# Missing Values
df.isnull().sum()

df.dropna()

df.fillna(0)

df.fillna(df.mean())

# Groupby
df.groupby("city")["salary"].mean()

# Sorting
df.sort_values("salary")

df.sort_values(
    "salary",
    ascending=False
)
```
